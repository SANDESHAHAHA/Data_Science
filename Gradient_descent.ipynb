```python
def logloss(y_true,y_actual_predictions):
  epsilon=1e-15
  y_predicted_new=np.clip(y_actual_predictions,epsilon,1-epsilon)
  return -np.mean(y_true*np.log(y_predicted_new)+(1-y_true)*np.log(1-y_predicted_new))
```

```python
#it gives the sigmoid of the entire array
def sigmoid_numpy(X):
  return 1/(1+np.exp(-X))

sigmoid_numpy(np.array([12,0,1]))
```

```python
def gradient_descent(age,affordability,y_true,epochs,loss_thresold=0.4631):
  w1=w2=1
  bias=0
  rate=0.1
  n=len(age)
  for i in range(epochs):
    weighted_sum= w1*age+w2*affordability+bias
    y_predicted=sigmoid_numpy(weighted_sum)
    loss=logloss(y_true,y_predicted)
    w1d=(1/n)*np.dot(np.transpose(age),(y_predicted-y_true))
    w2d=(1/n)*np.dot(np.transpose(affordability),(y_predicted-y_true))
    bias=np.mean(y_predicted-y_true)
    w1=w1-rate*w1d
    w2=w2-rate*w2d
    print(f'epoch:{i},w1:{w1},w2:{w2},bias:{bias},loss:{loss}')
    if loss<=loss_thresold:
      break
  return w1,w2,bias
```

```python
gradient_descent(X_train_scaled['age'],X_train_scaled['affordibility'],y_train,1000)
```
